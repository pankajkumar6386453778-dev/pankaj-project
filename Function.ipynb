##Generator function

```python
#Generator function
range(10)
```

```python
for i in range(10):
    print(i)
```

```python
#fibonacci number
def test_fib(n):
    a,b = 0,1
    for i in range(n):
        yield a # it means u used generator function
        a,b =b , a+b
```

```python
test_fib(10)
```

```python
for i in test_fib(10):
    print(i)
```

```python
def test_fib1():
    a,b = 0,1
    while True:
        yield a # it means u used generator function
        a,b =b , a+b

fib = test_fib1()
for i in range (10):
    print(next(fib))
```

```python
type(fib)
```

```python
def count_test(n):
    count =1 
    while count < n:
        yield count
        count = count +1
```

```python
c = count_test(5)
```
