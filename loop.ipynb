```python
#factorial
number = int(input("Enter your  number"))
factorial = 1
while number > 0 :
    factorial = factorial * number
    number = number-1
factorial
```

```python
#Fibonacci series
number = int(input("Enter the number the element you are looking for :"))
a ,b = 0 , 1
counter = 0

while counter < number :
    print(a)
    c = a+b
    a = b
    b = c
    counter = counter+1
```
